```kotlin
fun Map<Int, List<Int>>.existence() : MutableMap<Int, MutableList<Int?>> {
    val totalniExistence = mutableMapOf<Int, MutableList<Int?>>()

    for (x in 1..this.size) {
        for (y in 1..this.size) {

            /*
            println(this.get(x)) //list<Int>
            println(this.get(y)) //list<Int>
            */
            if (x != y) {
                if (hasEEEEE(this.get(x), this.get(y))) {
                    totalniExistence[x]?.add(y)
                }
            }
        }
    }

    val nesmiProvest = totalniExistence.values.any { it.isEmpty() }

    println("Selze: $nesmiProvest")
    return totalniExistence
}


fun hasEEEEE(f: List<Int>?, s: List<Int>?): Boolean {
    if (f == null || s == null) return false
    return s.any { it in f }
}

fun MutableMap<Int, MutableList<Int?>>.relace() : Int? {
    //vraci jako `vyberDalsi`
    println(this)
    val minimumVyskytu = mutableMapOf<Int, Int?>()
    for (x in 0..this.keys.size) {
        minimumVyskytu.put(x, this.get(x)?.size)
    }

    val nejnizsi = minimumVyskytu
        .filterValues { it != null }
        .minByOrNull { it.value!! }
        ?.key
    return nejnizsi
}


fun MutableMap<Int, MutableList<Int?>>.umisteni() : MutableMap<Int, MutableList<Int?>> {
    val umistiNejnizsiIndex = this.relace()
    val hodnota = this[umistiNejnizsiIndex]?.firstOrNull()

    println(umistiNejnizsiIndex)
    this.values.forEach { list -> list.remove(hodnota) }
    this.values.forEach { invalidRec -> invalidRec.remove(umistiNejnizsiIndex)}
    return this
}

//puzzle neni pridavana ale vytvari relaci a pote relaci k druhe puzzli  3 x 3 ^ 2
fun MutableList<Int?>?.relaceKPuzz(mapa: MutableMap<Int, MutableList<Int?>>, index: Int) {
    val listMoznosti = mutableMapOf<Int, MutableList<Int>>()

    //println(this)
    //println(this!!.size)

    if (this == null) return
    //musi ignorovat mapovani pro momentalni index (this = [x je 1], ignoruj mapa[1])
    for (x in 1..mapa.size) {
        if (x != index){
            val targetList = mapa[x]

            targetList?.forEach { v ->
                if (v != null && this.contains(v)) {
                    listMoznosti[index]?.add(x)
                }
            }
        }
    }
    println("[$index] -> ${listMoznosti[index]?.distinct()}")
    System.out.flush()
}


fun main() {
    val res = puzzles
        .existence()
        .umisteni()

    val a = res
    println("M $res")

    //relace k puzzli musi se zmenit pro vybrany list
    for (x in 1..a.size) {

        val list: MutableList<Int?>? = a.get(x)


        list.relaceKPuzz(a, x)
    }

    val kotlin = "🙂"
    println(kotlin)
}
```

```kotlin
fun pocitejVyskyt(mapa: Map<Int, List<Int>>): Map<Int, Int> {
    // Vytvoříme mutable mapu s předvyplněnými hodnotami 0..199 -> 0
    val vmap = (1..18).associateWith { 0 }.toMutableMap()

    // Procházíme jednotlivé seznamy v hodnotách mapy
    mapa.values.forEach { seznam ->
        seznam.forEach { cislo ->
            // Bezpečné přičtení (pouze pokud je číslo v rozsahu 0..199)
            if (cislo in vmap) {
                vmap[cislo] = vmap.getValue(cislo) + 1
            }
        }
    }

    return vmap
}

println(pocitejVyskyt(puzzles))
println("hi")
System.out.flush()
```
